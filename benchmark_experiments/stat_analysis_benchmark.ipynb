```python
import glob

import pandas as pd

UNCERTAINTY_TYPE = 'ALL'
METRICS = [  'PRR (MAE)']

path = "./results-ENS (LGBM)-ERROR/*.csv"

df_raw = pd.DataFrame()

for fname in glob.glob(path):

    if fname.endswith("-rejection-curve.csv") or fname.endswith("-pred-probas.csv") or fname.endswith(
            "-true-classes.csv"):
        continue

    print(fname)
    
    df = pd.read_csv(fname)

    df_raw = pd.concat([df_raw, df])

df_raw

print(df_raw[df_raw.isna().any(axis=1)])
df_raw = df_raw.fillna(0)

if not UNCERTAINTY_TYPE == 'ALL':
    df_raw = df_raw[df_raw['Uncertainty Type'] == UNCERTAINTY_TYPE]

df_raw = df_raw.sort_values(['Uncertainty Measure', 'Dataset'])
```

```python
if not UNCERTAINTY_TYPE == 'ALL':
    # Rename uncertainty measures
    df_raw = df_raw.replace(f'Entropy', f"$\\text{{{UNCERTAINTY_TYPE}}}_{{\\,\\text{{ent}}}}$")
    df_raw = df_raw.replace(f'Variance', f"$\\text{{{UNCERTAINTY_TYPE}}}_{{\\,\\text{{var}}}}$")
    df_raw = df_raw.replace(f"Binary Entropy", f"$\\text{{{UNCERTAINTY_TYPE}}}_{{\\,\\text{{ord-ent}}}}$")
    df_raw = df_raw.replace(f"Binary Variance", f"$\\text{{{UNCERTAINTY_TYPE}}}_{{\\,\\text{{ord-var}}}}$")
    df_raw = df_raw.replace(f"Binary Entropy One vs. Rest", f"$\\text{{{UNCERTAINTY_TYPE}}}_{{\\,\\text{{bin-ent}}}}$")
    df_raw = df_raw.replace(f"Binary Variance One vs. Rest", f"$\\text{{{UNCERTAINTY_TYPE}}}_{{\\,\\text{{bin-var}}}}$")
else:
    # Rename uncertainty measures
    df_raw = df_raw.replace("Entropy", "ent")
    df_raw = df_raw.replace("Variance", "var")
    df_raw = df_raw.replace("Binary Entropy", "ord-ent")
    df_raw = df_raw.replace("Binary Variance", "ord-var")
    df_raw = df_raw.replace(f"Binary Entropy One vs. Rest", "bin-ent")
    df_raw = df_raw.replace(f"Binary Variance One vs. Rest", "bin-var")


# Rename uncertainty measures
# df_raw = df_raw.replace("Entropy", "ent")
# df_raw = df_raw.replace("Variance", "var")
# df_raw = df_raw.replace("Binary Entropy", "ord-ent")
# df_raw = df_raw.replace("Binary Variance", "ord-var")
```

```python
df_raw.index.name = 'index'
df_raw['index'] = df_raw.index
df_raw
```

```python
i = 0
j = 0
while True:
    if j == len(df_raw):
        break
    df_raw.iloc[j, df_raw.columns.get_loc('index')] = i
    if i == 10:
        i = 0
    else:
        i += 1
        j += 1
df_raw['index'].unique()
df_raw
```

```python
import pandas as pd

df_raw = df_raw.copy()

MAXIMIZE_OUTCOME = True
```

```python
df = pd.DataFrame()

for m in METRICS:
    df_next = df_raw.pivot(
        index=['index', "Dataset", 'Uncertainty Type'],
        columns="Uncertainty Measure",
        values=m
    )

    df = pd.concat([df, df_next])

df
```

```python
from critdd import Diagram

# create a CD diagram from the Pandas DataFrame
diagram = Diagram(
    df.to_numpy(),
    treatment_names=df.columns,
    maximize_outcome=MAXIMIZE_OUTCOME
)

# inspect average ranks and groups of statistically indistinguishable treatments
diagram.average_ranks  # the average rank of each treatment
diagram.get_groups(alpha=.05, adjustment="holm")

# export the diagram to a file
diagram.to_file(
    "cd_result.tex",
    alpha=.05,
    adjustment="holm",
    reverse_x=False,
    axis_options={"title": UNCERTAINTY_TYPE + ' - ' + ', '.join(METRICS)
                  },
)
```
