```python
import pandas as pd
import seaborn as sns

TYPE = 'TU'
DATA = 'left_tail'
PRR = "PRR (MAE)"
GROUP = 'tail'

prrs = pd.read_csv(f'prrs_{DATA}.csv')

# Rename uncertainty measures
prrs = prrs.replace(f'Entropy', "ent")
prrs = prrs.replace(f'Variance', "var")
prrs = prrs.replace(f"Binary Entropy", "ord-ent")
prrs = prrs.replace(f"Binary Variance", "ord-var")
prrs = prrs.replace(f"Binary Entropy One vs. Rest", "bin-ent")
prrs = prrs.replace(f"Binary Variance One vs. Rest", "bin-var")

prrs
```

```python
prrs = prrs[(prrs["Uncertainty Type"] == TYPE) & (prrs["Group"] == GROUP)]
prrs
```

```python
prrs['rank'] = prrs['PRR (MAE)'].rank(ascending=False)

prrs = prrs.groupby(["Group","Uncertainty Type","Uncertainty Measure" ]).agg(
            ['mean'])[('rank','mean')].rank().sort_values()

#  # Export as a Booktabs LaTeX table
# latex_str = prrs.to_latex(
#     index=True
# )
#
# # Save to file
# with open("./test.tex", 'w') as f:
#     f.write(latex_str)
prrs
```

```python
from matplotlib import pyplot as plt



sns.set_theme(context='paper', style='whitegrid', font_scale=2.2, font="serif", rc={
    "text.usetex": True
})

ax = sns.boxplot(data=prrs, showfliers=False, x="Uncertainty Measure", y=PRR, hue="Uncertainty Measure")
ax.set(xlabel=None)

plt.savefig(TYPE + "_"+ DATA + "_"+GROUP+"_"+PRR+".pdf", bbox_inches='tight', dpi=300)

plt.show()
```
